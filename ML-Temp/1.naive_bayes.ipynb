```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('datasets/SalaryData_Train.csv')

df = df[['age', 'educationno', 'hoursperweek', 'Salary']]

y = df['Salary']
X = df.drop(columns=['Salary'])

X.dtypes
```

```python
classes = np.unique(y)
means = {}
variances = {}
priors = {}

for c in classes:
    X_c = X[y == c]
    means[c] = np.mean(X_c, axis=0)
    variances[c] = np.var(X_c, axis=0) + 1e-9
    priors[c] = X_c.shape[0] / X.shape[0]


def gaussian_pdf(x, class_val):
    mean = means[class_val]
    var = variances[class_val]
    numerator = np.exp(-(x-mean) ** 2 / (2 * var))
    denominator = np.sqrt(2 * np.pi * var)
    return numerator / denominator



def predict_single(x):
    posteriors = []
    for c in classes:
        prior = np.log(priors[c])
        likelihood = np.sum(np.log(gaussian_pdf(x, c)))
        posteriors.append((prior + likelihood, c))

    return max(posteriors)[1]


def predict(X):
    y_pred = [predict_single(x) for x in X]
    return np.array(y_pred)


y_pred = predict(X.to_numpy())
acc = np.mean(y.to_numpy() == y_pred)

acc
```
